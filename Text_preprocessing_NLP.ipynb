# Text Cleaning

```python
import re
```

```python
# نحفظ النص الخام داخل متغير حتى نستخدمه في جميع مراحل المعالجة القادمة
original_text = "<div>لَااا أُريدُ هذا المُنتَج!!! سعره 350 ريال، والتوصيل تأخــــر جِدًااا 😞https://store.com  </div> "

# نطبع عنوانًا يوضح أن النص التالي هو النص قبل المعالجة
print("النص الخام قبل المعالجة:")

# نطبع النص الأصلي كما كتبه المستخدم دون أي تعديل
print(original_text)
```

```python
# نبحث عن أي محتوى يبدأ بعلامة أصغر من وينتهي بعلامة أكبر من ثم نحذفه
text_without_html = re.sub(r"<[^>]+>", "", original_text)

# نعرض النص قبل إزالة وسوم صفحات الويب
print("النص قبل إزالة الوسوم:")
print(original_text)

# نضيف سطرًا فارغًا لتنظيم شكل النتيجة
print()

# نعرض النص بعد إزالة وسوم صفحات الويب
print("النص بعد إزالة الوسوم:")
print(text_without_html)
```

```python
# نبحث عن الروابط الإلكترونية داخل النص ونستبدلها بنص فارغ
text_without_urls = re.sub(
    r"https?://\S+|www\.\S+",
    "",
    text_without_html
)

# نعرض النص قبل إزالة الرابط الإلكتروني
print("النص قبل إزالة الرابط:")
print(text_without_html)

# نطبع سطرًا فارغًا لتنظيم عرض النتائج
print()

# نعرض النص بعد إزالة الرابط الإلكتروني
print("النص بعد إزالة الرابط:")
print(text_without_urls)
```

```python

text_without_punctuation = re.sub(
    r"""[!"#$%&'()*+,\-./:;<=>?@\[\]\\^_`{|}~،؛؟…]""",
    " ",
    text_without_urls
)

# نحذف الرموز الأخرى مثل الرموز التعبيرية مع الاحتفاظ بالحروف العربية والأرقام والمسافات
text_without_symbols = re.sub(
    r"[^\u0600-\u06FF\u0750-\u077F\u08A0-\u08FF0-9\s]",
    " ",
    text_without_punctuation
)

# نستبدل المسافات المتكررة بمسافة واحدة
text_without_symbols = re.sub(r"\s+", " ", text_without_symbols)

# نحذف المسافات الموجودة في بداية النص ونهايته
text_without_symbols = text_without_symbols.strip()

# نعرض النص قبل إزالة علامات الترقيم والرموز
print("النص قبل إزالة علامات الترقيم والرموز:")
print(text_without_urls)

# نطبع سطرًا فارغًا لتنظيم عرض النتيجة
print()

# نعرض النص بعد إزالة علامات الترقيم والرموز
print("النص بعد إزالة علامات الترقيم والرموز:")
print(text_without_symbols)
```

# Text normalization

```python
# ننسخ ناتج مرحلة التنظيف إلى متغير جديد لتطبيق التوحيد عليه
normalized_text = text_without_symbols

# نحذف علامات التشكيل مثل الفتحة والضمة والكسرة والشدة والتنوين
normalized_text = re.sub(
    r"[\u0617-\u061A\u064B-\u0652\u0670\u06D6-\u06ED]",
    "",
    normalized_text
)

# نحذف علامة التطويل الموجودة بين الحروف مثل هــــذا
normalized_text = re.sub(
    r"\u0640",
    "",
    normalized_text
)

# نوحد أشكال حرف الألف المختلفة ونجعلها ألفًا عادية
normalized_text = re.sub(
    r"[أإآٱ]",
    "ا",
    normalized_text
)

# نقلل تكرار الحرف العربي إذا تكرر ثلاث مرات أو أكثر إلى حرف واحد
normalized_text = re.sub(
    r"([\u0621-\u064A])\1{2,}",
    r"\1",
    normalized_text
)

# نستبدل المسافات المتكررة داخل النص بمسافة واحدة
normalized_text = re.sub(
    r"\s+",
    " ",
    normalized_text
)

# نحذف المسافات الموجودة في بداية النص ونهايته
normalized_text = normalized_text.strip()

# نعرض النص قبل تطبيق توحيد النص العربي
print("النص قبل التوحيد:")
print(text_without_symbols)

# نطبع سطرًا فارغًا لتنظيم عرض النتائج
print()

# نعرض النص بعد تطبيق توحيد النص العربي
print("النص بعد التوحيد:")
print(normalized_text)
```

# Tokenization

```python
# نقسم النص عند كل مسافة ونحفظ الكلمات الناتجة داخل قائمة
tokens = normalized_text.split()

# نعرض النص قبل تقسيمه إلى كلمات
print("النص قبل التقسيم:")

# نطبع النص بعد التنظيف والتوحيد
print(normalized_text)

# نطبع سطرًا فارغًا لتنظيم عرض النتائج
print()

# نعرض عنوان النتيجة بعد تقسيم النص
print("النص بعد التقسيم:")

# نطبع قائمة الكلمات الناتجة
print(tokens)

# نطبع عدد الوحدات النصية الموجودة في القائمة
print("عدد الوحدات النصية:", len(tokens))
```

# stop words

```python
# ننشئ مجموعة صغيرة من الكلمات العربية الشائعة التي نريد حذفها
stop_words = {
    "ان",
    "هذا",
    "هذه",
    "في",
    "من",
    "الى",
    "على",
    "عن",
    "هو",
    "هي",
    "كان",
    "كانت",
    "مع",
    "ثم",
    "او",
    "ا",
    "بهذا"
}

# ننشئ قائمة جديدة تحتوي على الكلمات غير الموجودة في قائمة الكلمات الشائعة
filtered_tokens = [
    token
    for token in tokens
    if token not in stop_words
]

# نعرض الكلمات قبل حذف الكلمات الشائعة
print("الكلمات قبل حذف الكلمات الشائعة:")
print(tokens)

# نطبع سطرًا فارغًا لتنظيم عرض النتائج
print()

# نعرض الكلمات بعد حذف الكلمات الشائعة
print("الكلمات بعد حذف الكلمات الشائعة:")
print(filtered_tokens)

# نعرض عدد الكلمات قبل الحذف
print("عدد الكلمات قبل الحذف:", len(tokens))

# نعرض عدد الكلمات بعد الحذف
print("عدد الكلمات بعد الحذف:", len(filtered_tokens))
```

# Stemming

```python
# نستورد أداة الاختزال الخاصة بالكلمات من مكتبة معالجة اللغة
from nltk.stem.snowball import SnowballStemmer

# ننشئ أداة اختزال مخصصة لمعالجة الكلمات العربية
stemmer = SnowballStemmer("arabic")

# نطبق الاختزال على الكلمات ونحتفظ بالأرقام كما هي
stemmed_tokens = [
    stemmer.stem(token) if token.isalpha() else token
    for token in filtered_tokens
]

# نعرض الكلمات قبل تطبيق الاختزال
print("الكلمات قبل الاختزال:")
print(filtered_tokens)

# نطبع سطرًا فارغًا لتنظيم عرض النتائج
print()

# نعرض الكلمات بعد تطبيق الاختزال
print("الكلمات بعد الاختزال:")
print(stemmed_tokens)

# نطبع سطرًا فارغًا قبل عرض المقارنة
print()

# نعرض كل كلمة بجانب الشكل الناتج عنها بعد الاختزال
print("مقارنة الكلمات قبل الاختزال وبعده:")

# نمر على الكلمات الأصلية والكلمات المختزلة معًا
for original_token, stemmed_token in zip(filtered_tokens, stemmed_tokens):

    # نطبع الكلمة الأصلية ثم الشكل المختزل منها
    print(original_token, "←", stemmed_token)
```

# Lemmatization

```python
# نثبت مكتبة ستانزا داخل بيئة كولاب
!pip install -q stanza

# نستورد مكتبة ستانزا لمعالجة اللغة
import stanza

# نحمل النموذج العربي اللازم للتقسيم وتحديد نوع الكلمة والإرجاع إلى الصيغة الأساسية
stanza.download(
    "ar",
    processors="tokenize,pos,lemma",
    verbose=False
)

# ننشئ خط معالجة مخصصًا للغة العربية
# نوضح أن الكلمات مقسمة مسبقًا حتى لا تعيد الأداة تقسيمها
lemmatizer = stanza.Pipeline(
    lang="ar",
    processors="tokenize,pos,lemma",
    tokenize_pretokenized=True,
    use_gpu=False,
    verbose=False
)

# نرسل الكلمات الناتجة بعد حذف الكلمات الشائعة إلى أداة المعالجة
document = lemmatizer([filtered_tokens])

# ننشئ قائمة فارغة لحفظ الصيغ الأساسية للكلمات
lemmatized_tokens = []

# نمر على جميع الجمل التي أعادتها أداة المعالجة
for sentence in document.sentences:

    # نمر على كل كلمة داخل الجملة
    for word in sentence.words:

        # نحتفظ بالصيغة الأساسية إذا وجدتها الأداة وإلا نحتفظ بالكلمة الأصلية
        lemma = word.lemma if word.lemma else word.text

        # نضيف الصيغة الأساسية إلى القائمة
        lemmatized_tokens.append(lemma)

# نعرض الكلمات قبل تطبيق الإرجاع إلى الصيغة الأساسية
print("الكلمات قبل تطبيق (Lemmatization):")
print(filtered_tokens)

# نطبع سطرًا فارغًا لتنظيم عرض النتائج
print()

# نعرض الكلمات بعد تطبيق الإرجاع إلى الصيغة الأساسية
print("الكلمات بعد تطبيق (Lemmatization):")
print(lemmatized_tokens)

# نطبع سطرًا فارغًا قبل عرض المقارنة
print()

# نعرض عنوان المقارنة
print("مقارنة الكلمات قبل وبعد (Lemmatization):")

# نمر على الكلمات الأصلية والصيغ الأساسية معًا
for original_token, lemma_token in zip(filtered_tokens, lemmatized_tokens):

    # نطبع الكلمة الأصلية بجانب الصيغة الأساسية التي أعادتها الأداة
    print(original_token, "←", lemma_token)
```

# Part-of-Speech Tagging (POS Tagging)

```python
# نرسل الكلمات بعد التنظيف والتوحيد إلى أداة التحليل اللغوي
pos_document = lemmatizer([tokens])

# ننشئ قاموسًا لتحويل الاختصارات إلى أسماء عربية واضحة
pos_names = {
    "NOUN": "اسم",
    "VERB": "فعل",
    "ADJ": "صفة",
    "ADV": "ظرف",
    "PRON": "ضمير",
    "DET": "أداة تعريف",
    "ADP": "حرف جر",
    "CONJ": "حرف عطف",
    "CCONJ": "حرف عطف",
    "SCONJ": "أداة ربط",
    "PART": "أداة",
    "NUM": "رقم",
    "PROPN": "اسم علم",
    "AUX": "فعل مساعد",
    "PUNCT": "علامة ترقيم"
}

# ننشئ قائمة فارغة لحفظ الكلمات وأنواعها
pos_results = []

# نمر على جميع الجمل الناتجة من أداة التحليل
for sentence in pos_document.sentences:

    # نمر على كل كلمة داخل الجملة
    for word in sentence.words:

        # نحصل على الاسم العربي لنوع الكلمة
        arabic_pos = pos_names.get(word.upos, word.upos)

        # نضيف الكلمة ونوعها إلى قائمة النتائج
        pos_results.append((word.text, arabic_pos, word.upos))

# نعرض عنوان النتيجة
print("نتيجة تحديد نوع الكلمة (POS Tagging):")

# نطبع سطرًا فارغًا لتنظيم المخرجات
print()

# نعرض كل كلمة بجانب نوعها
for word, arabic_pos, pos_code in pos_results:

    # نطبع الكلمة ونوعها العربي والاختصار العالمي
    print(word, "←", arabic_pos, f"({pos_code})")
```
