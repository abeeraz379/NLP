# Regular Expression Methods in Python

Re Module + Regex Methods

Function

- findall : Returns a list containing all matches
- search : Returns a Match object if there is a match anywhere in the string
- split : Returns a list where the string has been split at each match
- sub : Replaces one or many matches with a string




```python
import re
```

```python
qoute="there's only one thing I hate more than laying : skim milk. which is water that's laying to milk. -Ron swsan "
```

# Search

```python
re.search('milk',qoute)
```

give is one of th matches (search)

```python
re.search('milk',qoute).group()
```

# Findall

```python
re.findall('milk',qoute)
```

```python
len(re.findall('milk',qoute))
```

```python
re.findall('[a-m]',qoute)
# find all ch from a to m
```

```python
re.findall('[a-cA-C]',qoute)
# find all ch from a to c capital and small
```

```python
s ='abcdfghigk'
re.findall('.',s)
```

```python
re.findall('a.c',s)
```

# Split

```python
re.split('milk',qoute)
```

```python
re.split('\.',qoute)
```

# Sub

```python
re.sub("I","You",qoute,count=1)
```

count = 1 is to replace the first one

```python
re.subn("a","b",s)
```
